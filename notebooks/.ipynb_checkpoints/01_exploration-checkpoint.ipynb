```python
import pandas as pd
customers = pd.read_csv("../seeds/raw_customers.csv")
orders = pd.read_csv("../seeds/raw_orders.csv",
parse_dates=["ordered_at"])
items = pd.read_csv("../seeds/raw_items.csv")
products = pd.read_csv("../seeds/raw_products.csv")
supplies = pd.read_csv("../seeds/raw_supplies.csv")
stores = pd.read_csv("../seeds/raw_stores.csv",
parse_dates=["opened_at"])
tables = {
"customers": customers, "orders": orders, "items": items,
"products": products, "supplies": supplies, "stores": stores,
}
for name, df in tables.items():
    print(f"{name}: {df.shape[0]} lignes, {df.shape[1]} colonnes")
```

```python
for name, df in tables.items():
    print(f"--- {name} ---")
    print(df.dtypes)
    print("Valeurs manquantes :")
    print(df.isnull().sum())
    print()
```

```python
print("customers : id est-il unique ?", customers["id"].is_unique)
print(customers.head())
print()
print("stores : id est-il unique ?", stores["id"].is_unique)
print(stores)
```

```python
print("orders : id est-il unique ?", orders["id"].is_unique)
print("Plage de dates :", orders["ordered_at"].min(), "->",
 orders["ordered_at"].max())
print("Nombre de clients distincts :",
 orders["customer"].nunique())
print("Nombre de magasins distincts :",
 orders["store_id"].nunique())
orders[["subtotal", "tax_paid", "order_total"]].describe()
```

```python
print("items : id est-il unique ?", items["id"].is_unique)
print("Nombre d'order_id distincts references par items :",
 items["order_id"].nunique())
print("Nombre de commandes dans raw_orders :",
 orders["id"].nunique())
orphan_check = set(items["order_id"]) - set(orders["id"])
print("order_id dans items mais absents de orders :",
 len(orphan_check))
```

```python
print(products[["sku", "name", "type", "price"]])
print()
print(supplies[["id", "name", "cost", "perishable",
 "sku"]].head(10))
```

```python
product_skus = set(products["sku"])
supply_skus = set(supplies["sku"].unique())
item_skus = set(items["sku"])
print("SKU des items references dans products :", len(item_skus &
 product_skus))
print("SKU des items references dans supplies :", len(item_skus &
 supply_skus))
print("SKU des items non references nulle part :", len(item_skus -
 product_skus - supply_skus))
```
